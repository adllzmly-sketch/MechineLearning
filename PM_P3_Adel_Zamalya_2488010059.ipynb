# Latihan 1

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style='whitegrid')
df = sns.load_dataset('tips')   # butuh internet (Colab)
df.head()
df.shape
df.info()
```

# Latihan 2

```python
# Histogram total_bill
plt.figure(figsize=(7, 4))
sns.histplot(df['total_bill'], bins=20, kde=True)
plt.title('Distribusi Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frekuensi')
plt.show()

# Box plot tip
plt.figure(figsize=(7, 3))
sns.boxplot(x=df['tip'])
plt.title('Box Plot Tip')
plt.xlabel('Tip')
plt.show()
```

Pada histogram total_bill, sebagian besar tagihan berada di kisaran 10–20. Ada beberapa tagihan yang nilainya lebih tinggi sehingga sebaran datanya lebih banyak ke nilai yang besar. Pada box plot tip, sebagian besar tip berada di kisaran 2–4, dan terdapat beberapa nilai yang jauh lebih tinggi yang termasuk outlier.

# Latihan 3

```python
# Scatter plot total_bill vs tip
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x='total_bill', y='tip', hue='time')

plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

Dari scatter plot dapat disimpulkan bahwa semakin besar total tagihan, biasanya jumlah tip juga semakin besar. Data Lunch dan Dinner memiliki jumlah tagihan yang beragam, tetapi kebanyakan masih berada di kisaran yang rendah.

# Latihan 4

```python
df_numerik = df.select_dtypes(include='number')

plt.figure(figsize=(7, 5))
sns.heatmap(df_numerik.corr(), annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Heatmap Korelasi Dataset Tips')
plt.show()
```

Berdasarkan heatmap tersebut, pasangan fitur yang paling berkorelasi adalah total_bill dan tip dengan nilai korelasi sekitar 0,68.

# 1. Mengapa EDA penting dilakukan sebelum membangun model?

EDA penting karena untuk membuat model, kita perlu memahami data terlebih dahulu. Dengan EDA, kita bisa mengetahui bentuk data, melihat pola, menemukan data yang tidak biasa atau outlier, serta mengetahui hubungan antar fitur.

# 2. Apa perbedaan informasi yang Anda peroleh dari histogram, box plot, dan scatter plot?

Histogram biasanya untuk melihat sebaran data. Box plot itu untuk melihat sebaran dan outlier, sedangkan scatter plot untuk melihat hubungan antara dua data.
